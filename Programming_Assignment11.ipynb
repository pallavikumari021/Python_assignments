1. Write a Python program to find words which are greater than given length k?

```python
def word_k(k, s):
    word = s.split(" ")
    for x in word:
        if len(x)>k:
            print(x)

k = 3
s = "Hello world"
word_k(k, s)
```

2. Write a Python program for removing i-th character from a string?

```python
def remove_char(s, i):
    a = s[ : i]
    b = s[i + 1: ]

    return a+b

string = "Pythonisgood"
i = 5
print(remove_char(string, i-1))
```

3. Write a Python program to split and join a string?

```python
def splitting(str):
    list = str.split(' ')
    return list
def joining(list):
    str = ' '.join(list)
    return str
str = 'Once upon a time'
list = splitting(str)
print("The string after splitting is: ", list)

restoring = joining(list)
print("The string after joining is: ", restoring)
```

4. Write a Python to check if a given string is binary string or not?

```python
def check(string) :
    b = set(string)
    s= {'0', '1'}
    if s == b or b == {'0'} or b == {'1'}:
        print("Binary String")
    else:
        print("Non Binary String")
        
s1 = "00110101"
check(s1)
s2 = "1010100200111"
check(s2)
```

5. Write a Python program to find uncommon words from two Strings?

```python
def uncommon_words(s1, s2):
    count = {}
    for word in s1.split():
        count[word] = count.get(word, 0) + 1
    for word in s2.split():
        count[word] = count.get(word, 0) + 1
    return [word for word in count if count[word] == 1]

s1 = "Hello World"
s2 = "Hello World ineuron"

print(uncommon_words(s1, s2))
```

6. Write a Python to find all duplicate characters in string?

```python
def duplicate_characters(string):
    chars = {}

    for char in string:
        if char not in chars:
            chars[char] = 1
        else:
            chars[char] += 1

    duplicates = []

    for char, count in chars.items():
        if count > 1:
            duplicates.append(char)
            
    return duplicates
print(duplicate_characters("hello"))
```

```python
import re
s = "Hell$o"
def has_special_char(s):
   pattern = r'[^a-zA-Z0-9\s]' 
   output = bool(re.search(pattern, s))
   if output == True:
      print(s, "has the special characters in it") 
   else:
      print(s, "has no special characters in it")
has_special_char(s)
```
